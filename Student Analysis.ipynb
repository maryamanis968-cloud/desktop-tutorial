# Pie Chart(Male vs Female students)

```python
# gender column ka count nikal kar pie chart ban dia
gender_count = df["gender"].value_counts()

plt.pie(gender_count.values,labels = gender_count.index,autopct='%1.1f%%',startangle = 90)

plt.title("Male vs Female Students")
plt.show()
        
```

# Box Plot(CGPA distribution by gender)

```python
# cgpa kis range ke beech mai hai

plt.figure(figsize=(7,4))

sns.boxplot(data=df, x="gender", y="cgpa")

plt.title("CGPA Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("CGPA")

plt.show()
```

```python
# groupby karwa kar har gender ka cgpa average nikal liya
df.groupby("gender")["cgpa"].mean()
```

# Bar Plot(average cgpa by gender)

```python
# average cgpa kis gender ka zyada hai

avg_cgpa = df.groupby("gender")["cgpa"].mean()

plt.figure(figsize=(6,6))
plt.bar(avg_cgpa.index,avg_cgpa.values, color="orange")   # bar chart
plt.title("Average CGPA by Gender")
plt.xlabel("Gender")
plt.ylabel("Average CGPA")
plt.show()
```

# Histogram:
## Communication Skill Score

```python
# mostly students ka score 60 se 80 ke beech hai

plt.figure(figsize=(7,4))
plt.hist(df["communication_skill_score"], bins=10, color="skyblue", edgecolor="black") # eq single numeric values ka distribution
# bins=10 matlab data ko 10 groups (bins) mein divide kiya
plt.title("Communication Skill Score - Histogram")
plt.xlabel("Score (Binned)")
plt.ylabel("Frequency") # har score range mai kitne students hai
plt.show()
```

# Heat Map

```python
# 0.00 mtlb ke is dataset ke according in scores ke darmian linear relation/corr almost nhi hai,aur diagnol per 1 isliye hai koike har variable
# ka khud se corelation 1 hota hai

corr = df[["mock_interview_score","aptitude_score","communication_skill_score"]].corr() # do cheezon ke darmian relationship

plt.figure(figsize=(5,4))
sns.heatmap(corr, annot=True, cmap="YlGnBu", fmt=".2f")
# annot=True -> har cell mein number bhi dikhega
plt.title("Correlation Heatmap (Seaborn)")
plt.show()
```

# Internship Count by Average Salary

```python
# har internship count ke liye average salary package nikal rha hai
df.groupby("internships_count")["salary_package_lpa"].mean()
```

```python
import matplotlib.pyplot as plt

# 1. Initialize figure size BEFORE plotting
plt.figure(figsize=(10, 5))

# 2. Generate the plot (kind='bar' as a string)
df.groupby("internships_count")["salary_package_lpa"].mean().plot(
    kind="bar", color="skyblue"
)

# 3. Add titles and labels matching your data
plt.title(
    "Average Salary Package by Internships Count", fontsize=16, fontweight="bold"
)
plt.xlabel("Number of Internships", fontsize=12)
plt.ylabel("Average Salary Package (LPA)", fontsize=12)
plt.xticks(rotation=0)  # Keep x-axis labels readable

# 4. Display the plot
plt.show()
```

# Attendace Percentage By Branch

```python
# ye har branch ke liye average attendance % nikal rha hai
df.groupby("branch")["attendance_percentage"].mean()
```

```python
import matplotlib.pyplot as plt

# 1. Initialize figure size BEFORE plotting
plt.figure(figsize=(10, 5))

# 2. Generate the plot (kind='bar' as a string)
df.groupby("branch")["attendance_percentage"].mean().plot(
    kind="bar", color="skyblue"
)

# 3. Add titles and labels matching your data
plt.title(
    "Attendance Percentage by Branch", fontsize=16, fontweight="bold"
)
plt.xlabel("Branch", fontsize=12)
plt.ylabel("Attendane Percentage", fontsize=12)
plt.xticks(rotation=0)  # Keep x-axis labels readable

# 4. Display the plot
plt.show()
```
