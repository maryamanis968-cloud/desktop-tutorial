```python
class Patient:
    def __init__(self,patient_id,name,age,gender,disease):
        self.patient_id = patient_id
        self.name = name
        self.age = age
        self.gender = gender
        self.disease = disease
        
    def display(self):
        print(f"Patient Details :{self.patient_id}, {self.name}, {self.age}, {self.gender}, {self.disease}" )
    
```

```python
class Hospital:

    def __init__(self):
        self.patients = []

    def add_patient(self, name, age, disease):
        patient = {"name": name, "age": age, "disease": disease}
        self.patients.append(patient)
        print(f"Patient '{name}' added successfully!")

    def show_patient(self):
        if not self.patients:
            print("No patients found in the system.")
            return

        print("\n--- Patient List ---")
        for number, p in enumerate(self.patients, 1):#iska matlb hai ke number 1 se patient ki indexing shuru karo..enumerate se no mile ga warna nhi
            print(
                f"{number}. Name: {p['name']} | Age: {p['age']} | Disease:"
                f" {p['disease']}"
            )


# Usage:
hospital = Hospital()

# Patient add karte hain
hospital.add_patient("Ali", 28, "Fever")
hospital.add_patient("Sara", 35, "Headache")

# Patient ki details show karne ke liye:
hospital.show_patient()
```

```python
class Person:
    def __init__(self,name,age,height):
        self.name = name
        self.age = age
        self.height = height

    def show(self):
        print(f" {self.name} , {self.age} , {self.height}" )

    
```

```python
class Manager(Person):
    def __init__(self,name,age,height,manager_id,wardno,shift):
        super().__init__(name,age,height)
        self.name = name 
        self.age = age
        self.height = height
        self.manager_id = manager_id
        self.wardno = wardno
        self.shift = shift

    def show_manager(self):
        print(f"Manager:Name: {self.name} | Age: {self.age} | Height: {self.height} | Manager_id: {self.manager_id} | Wardno: {self.wardno} |Shift: {self.shift}" )
```

```python
class Patient(Person):
    def __init__(self,name,age,height,patient_id,disease,wardno):
        super().__init__(name,age,height)
        self.name = name
        self.age  = age
        self.height = height
        self.patient_id = patient_id
        self.disease = disease
        self.wardno = wardno

    def show_patient(self):
        print(f"Patient:Name: {self.name} |Age: {self.age} | Height: {self.height}| Patient_id: {self.patient_id} |Disease: {self.disease} |Wardno: {self.wardno}" )

        
```

```python
m1 = Manager("Ali",35,"5.8 ft",102,2,"Morning")
        
p1 = Patient("Ahmed",25,"5.6 ft",201,"Fever",2)

m1.show_manager()
print()
p1.show_patient()

```

```python
# DUCK TYPING

class Doctor:
    def treat(self):
        print("Doctor is treating the patient: ")


class Nurse:
    def treat(self):
        print("Nurse is assisting the patient: ")


class Patient:
    def receive_care(self,person):
        person.treat()
        
p1 = Patient()
d1 = Doctor()
n1 = Nurse()

p1.receive_care(d1)
p1.receive_care(n1)
```
