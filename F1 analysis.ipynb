# Drivers Ratio
## Pie Chart

```python
import plotly.graph_objects as go

driver_data = df["Driver"].value_counts()

# 0.08 = halka separation
# 0.12 = zyada separation
pull = [0.08] * len(driver_data)

# Pie chart create kar rahe hain.
fig = go.Figure(
    go.Pie(
        labels=driver_data.index,
        values=driver_data.values,

        # Percentage + driver dono outside show honge.
        textinfo="label+percent",

        # Text ko slice ke bahar rakhta hai.
        textposition="outside",

        # Har slice ko center se thora bahar pull karta hai.
        pull=pull,

        # Percentage ko clean format mein show karta hai.
        texttemplate="%{label}<br>%{percent}",

        # Text readable rakhta hai.
        textfont=dict(
            size=14
        ),

        # Slices ke darmiyan clear white separation.
        marker=dict(
            line=dict(
                color="white",
                width=2
            )
        ),

        # Hover karne par additional information.
        hovertemplate=(
            "<b>%{label}</b><br>"
            "Count: %{value}<br>"
            "Percentage: %{percent}"
            "<extra></extra>"
        )
    )
)

# Poore chart ka size control kar rahe hain.
# width = chart ki horizontal size
# height = chart ki vertical size
fig.update_layout(
    width=900,
    height=900
)

# Chart ka overall layout.
fig.update_layout(
    title=dict(
        text="Drivers Ratio",
        x=1
    ),

    # Legend side par show hoga.
    showlegend=True
)

fig.show()
```

# Yearly Trend Records
## line plot

```python
import matplotlib.pyplot as plt

# 1. Sirf Year ka value count lein
year_counts = df["Year"].value_counts().sort_index()  # Sort karne se saal sequence mein aa jayenge

plt.figure(figsize=(7, 4))
plt.plot(year_counts.index, year_counts.values, color="teal", marker="o", linewidth=2)

plt.title("Yearly Trend")          # Graph ka sahi title
plt.xlabel("Year")                 # X-axis label
plt.ylabel("Count")                # Y-axis label
plt.grid(True, linestyle="--", alpha=0.6)
plt.show()
```

# Average Tyre Life by Compound
## barplot

```python
import matplotlib.pyplot as plt
import pandas as pd

# Agar har Compound ki average TyreLife nikalni ho:
avg_tyre_life = df.groupby("Compound")["TyreLife"].mean().reset_index() # groupby ke baad index banne walay column ko dobara normal column krne keliye.

plt.figure(figsize=(7, 4))
plt.bar(avg_tyre_life["Compound"], avg_tyre_life["TyreLife"], color="orange")

plt.title("Average Tyre Life by Compound")  
plt.xlabel("Compound")                       
plt.ylabel("Average Tyre Life")              
plt.xticks(rotation=45)


plt.show()
```
