# Room Type Ratio
## go.Pie

```python
df["room type"].value_counts()
```

```python
import plotly.graph_objects as go

# Har room type ki frequency/count nikal rahe hain.
# Yani har room type dataset mein kitni baar aya.
room_type = df["room type"].value_counts()

# Har slice ko thora sa separate karne ke liye.
# 0.08 = halka separation
# 0.12 = zyada separation
pull = [0.08] * len(room_type)

# Pie chart create kar rahe hain.
fig = go.Figure(
    go.Pie(
        labels=room_type.index,
        values=room_type.values,

        # Percentage + room type dono outside show honge.
        textinfo="label+percent",

        # Text ko slice ke bahar rakhta hai.
        textposition="outside",

        # Har slice ko center se thora bahar pull karta hai.
        pull=pull,

        # Percentage ko clean format mein show karta hai.
        texttemplate="%{label}<br>%{percent}",

        # Text readable rakhta hai.
        textfont=dict(
            size=14
        ),

        # Slices ke darmiyan clear white separation.
        marker=dict(
            line=dict(
                color="white",
                width=2
            )
        ),

        # Hover karne par additional information.
        hovertemplate=(
            "<b>%{label}</b><br>"
            "Count: %{value}<br>"
            "Percentage: %{percent}"
            "<extra></extra>"
        )
    )
)

# Poore chart ka size control kar rahe hain.
# width = chart ki horizontal size
# height = chart ki vertical size
fig.update_layout(
    width=700,
    height=700
)

# Chart ka overall layout.
fig.update_layout(
    title=dict(
        text="Room Type Ratio",
        x=0.5
    ),

    # Legend side par show hoga.
    showlegend=True
)

fig.show()
```

# Meal Type Ratio
## go.Pie()

```python
import plotly.graph_objects as go

meal_type = df["type of meal"].value_counts()

# Har slice ko thora sa separate karne ke liye
pull = [0.10] * len(meal_type)

# Pie chart create kar rahe hain
fig = go.Figure(
    go.Pie(
        labels=meal_type.index,
        values=meal_type.values,
        textinfo="label+percent",
        textposition="outside",
        pull=pull,

        texttemplate="%{label}<br>%{percent}",

        textfont=dict(
            size=14
        ),

        marker=dict(
            line=dict(
                color="black",
                width=0
            )
        ),

        hovertemplate=(
            "<b>%{label}</b><br>"
            "Count: %{value}<br>"
            "Percentage: %{percent}"
            "<extra></extra>"
        )
    )
)

# Chart ka overall size
fig.update_layout(
    width=700,
    height=700
)

# Chart ka title aur legend
fig.update_layout(
    title=dict(
        text="Meal Type Ratio",
        x=0.5
    ),
    showlegend=True
)

fig.show()
```
